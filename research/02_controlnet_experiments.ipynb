# AniQR - Phase 2: ControlNet & Anime Diffusion Experiments
This notebook conducts experiments with ControlNet conditioning scales (1.1 to 1.5), anime style presets, and automated OpenCV scannability validation.

```python
from aniqr_ai.pipeline.anime_qr_pipeline import AnimeQRPipeline
from aniqr_ai.components.anime_image_generation import AnimeGenerationParams, ANIME_STYLES

print('Available Anime Styles:')
for style_name, style_data in ANIME_STYLES.items():
    print(f' - {style_name}: {style_data["prompt"][:60]}...')
```

```python
# Run Pipeline with Studio Ghibli style
pipeline = AnimeQRPipeline()
params = AnimeGenerationParams(
    style='studio_ghibli',
    controlnet_conditioning_scale=1.25,
    guidance_scale=7.5,
    num_inference_steps=30,
)

output = pipeline.generate(
    value='https://github.com/aniqr',
    content_type='url',
    params=params,
)

print('Scannable:', output.is_scannable)
print('Attempts:', output.attempts)
print('Decoded:', output.decoded_content)
output.anime_qr_image
```
