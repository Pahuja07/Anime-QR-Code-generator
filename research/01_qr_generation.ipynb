# AniQR - Phase 1: QR Code Generation Engine
This notebook demonstrates generating standard and artistic base QR codes with error correction configuration (L, M, Q, H), click-to-call formats (`tel:`), URLs, and resolution sizing.

```python
from aniqr_ai.components.qr_generation import QRCodeGenerator, QRGenerationConfig, normalize_content
from aniqr_ai.components.qr_validation import QRValidator

# 1. Test normalization
print('URL:', normalize_content('github.com/aniqr', 'url'))
print('Text:', normalize_content('Hello Anime QR', 'text'))
print('Phone:', normalize_content('+1 (800) 555-0199', 'phone'))
```

```python
# 2. Generate High Error-Correction (Level H) QR code with 512x512 resolution
config = QRGenerationConfig(error_correction='H', box_size=10, border=4, target_size=(512, 512))
generator = QRCodeGenerator(config)

qr_img = generator.generate('https://aniqr.ai', 'url')
print('Generated QR Size:', qr_img.size)
qr_img
```

```python
# 3. Verify Scannability with OpenCV
validator = QRValidator()
result = validator.validate(qr_img, expected_content='https://aniqr.ai')
print('Scannable:', result.is_valid)
print('Decoded Text:', result.decoded_text)
```
